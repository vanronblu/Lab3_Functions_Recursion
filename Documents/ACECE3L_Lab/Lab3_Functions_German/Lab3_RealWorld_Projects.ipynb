```python
# Setup
LAST_NAME = "GERMAN"
SEED_NUM = 6
FAVORITE_ARTIST = "SHIREBOUND"

print("Student variables initialized successfully")
```

```python
#Exercise 1
def log_diagnostic(func):
    def wrapper(*args, **kwargs):
        print(f"[LOG] Running: {func.__name__}...")
        return func(*args, **kwargs)
    return wrapper

# Validation
def validate_readings(readings):
    valid = []
    invalid = 0 
    for r in readings:
        if isinstance(r, (int, float)) and 0 <= r <= 100:
            valid.append(r)
        else:
            invalid += 1
            print(f"    Invalid entry detected: {r}")
    return valid, invalid

# Calcu func using *args
def calculate_stats(*readings):
    if not readings:
        return 0.0 
    return sum(readings) / len(readings)

# Classification func
def classify_condition(avg):
    if avg >= 85:
        return "Optimal"
    elif avg >= 60:
        return "Warning"
    else:
        return "Critical"

# Main
@log_diagnostic
def run_diagnostics(*readings):
    # Validate
    valid_data, invalid_count = validate_readings(readings)

    if not valid_data:
        return {"status": "Failed", "reason": "No valid data"}

    # Calculate
    avg = calculate_stats(*valid_data)

    # Classify
    status = classify_condition(avg)
    return {"valid_count": len(valid_data), "invalid_count": invalid_count, "average": avg, "status": status}

data_payload = [
     len(LAST_NAME) * 10,
    SEED_NUM * 10,
    len(FAVORITE_ARTIST) * 5,
    "Sensor_Error"
]

# Execute Stage
results = run_diagnostics(*data_payload)

# Display output
print()
print("    Diagnostic Summary: ")
print(f"Valid Readings Processed: {results['valid_count']}")
print(f"Invalid Readings Skipped: {results['invalid_count']}")
print(f"Average Output: {results['average']:.2f}")
print(f"Final Equipment Condition: {results['status']}")
```

```python
# Exercise 2
trace_count = 0
def trace_fault(code, step=1):
    global trace_count
    trace_count +=1

    # Base Case
    if code <= 0:
        print(f"Step {step}: Reached base condition (Code: {code}). Stopping trace.")
        return "Resolved"

    # Recursive Step
    print(f"Step {step}: Tracing fault code {code}...")
    next_code = code - SEED_NUM

    return trace_fault(next_code, step + 1)

# Student specific fault code
# Last name length * seed_num, + a base value
initial_fault_code = (len(LAST_NAME) *SEED_NUM) + 10 # = 46

print(f"Initial Fault Code Generated: {initial_fault_code}")
print()
print("    RECURSIVE TRACE START ")

# Execute recursion
final_status = trace_fault(initial_fault_code)

print()
print(f"    TRACE COMPLETE ")
print(f"Total Recursive Calls Made: {trace_count}")
print(f"Final Result: {final_status}")
```

```python
# Exercise 3
def safe_read(readings):
    """Generator that yield valid readings and catches errors."""
    for r in readings:
        try:
            # Simulate data
            if not isinstance(r, (int, float)) or r < 0:
                raise ValueError(f"Invalid sensor reading: {r}")
            yield r
        except ValueError as e:
            print(f"[Error] {e}")

def analyze_anomaly(val, depth=0):
    """Recusive function to analyze a specific reading."""
    if depth >= 3 or val == 0: #prevent inifinite recursion
        return "Stable"

    # Simulate a recursive check
    return analyze_anomaly(val / 2, depth + 1)

# Decorator
def monitor_pipeline(func):
    def wrapper(*args, **kwargs):
        print("[Monitor] Initializing processing pipeline...")
        result = func(*args, **kwargs)
        print("[Monitor] Pipeline execution finished.")
        print()
        return result
    return wrapper

# Main Pipeline
@monitor_pipeline
def process_telemetry_stream(data_stream):
    valid_count = 0
    invalid_count = 0
    anomalies = 0
    total_sum = 0

    # Generator processes stream lazily
    for reading in safe_read(data_stream):
        valid_count += 1

        # Lambda transforms thee data
        scaled_reading = (lambda x: x * 1.5)(reading)

        # Check for anomalies using recursion
        status = analyze_anomaly(scaled_reading)
        if status != "Stable" or scaled_reading > 80:
            anomalies += 1

            total_sum += scaled_reading

    # Calculate final metrics
    avg = total_sum / valid_count if valid_count > 0 else 0
    final_status = "Optimal" if anomalies == 0 else "Needs Maintenance"

    return {
        "valid": valid_count,
        "invalid": len(data_stream) - valid_count,
        "anomalies": anomalies,
        "avg": avg,
        "status": final_status

    } 

# Relemetry Stream
# Mix of valid # invalid readings to test pipeline
raw_stream = [
    len(LAST_NAME) * 10,
    SEED_NUM * 2,
    len(FAVORITE_ARTIST) * 5,
    -50,
    "Null",
    10
]

# Execute stage
report = process_telemetry_stream(raw_stream)

# Final
print("   Final Diagnostic Report")
print(f"Total Processed Readings : {len(raw_stream)}")
print(f"Valid Readings           : {report['valid']}")
print(f"Invalid Readings         : {report['invalid']}")
print(f"Detected Anomalies       : {report['anomalies']}")
print(f"Average Telemetry        : {report['avg']:.2f}")
print(f"Overall Equipment Status : {report['status']}")
```
